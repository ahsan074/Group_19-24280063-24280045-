```python
import praw
import csv
import os
import datetime

# Define Reddit API Credentials
REDDIT_CLIENT_ID = "CENSORED"
REDDIT_CLIENT_SECRET = "CENSORED"
REDDIT_USER_AGENT = "CENSORED"

# Initialize Reddit API
reddit = praw.Reddit(
    client_id=REDDIT_CLIENT_ID,
    client_secret=REDDIT_CLIENT_SECRET,
    user_agent=REDDIT_USER_AGENT,
)

# Check if authentication works
try:
    print(f"✅ Authenticated as: {reddit.user.me()}")
except Exception as e:
    print(f"❌ Authentication failed: {e}")

# Search mechanism for Reddit
subreddit = reddit.subreddit("AskDocs+medicine+medical+healthcare")  # Targeting healthcare-related subreddits
search_query = "cancer OR diabetes OR surgery OR insurance OR hospital OR treatment OR patient"

# Fetch healthcare-related posts
posts = []
for submission in subreddit.search(search_query, limit=100):
    posts.append([
        submission.title,  # Post Title (Medical Condition)
        submission.author.name if submission.author else "Unknown",  # Author (Doctor/Source)
        datetime.datetime.fromtimestamp(submission.created_utc).isoformat(),  # Date of Admission (Timestamp)
        submission.subreddit.display_name,  # Subreddit (Hospital/Category)
        submission.score,  # Upvotes (Relevance)
        submission.num_comments,  # Comments (Discussion Count)
        submission.url  # URL (Reference)
    ])

# Save to CSV in Colab
csv_filename = "/content/reddit_healthcare_extracted.csv"
with open(csv_filename, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Medical Condition", "Doctor", "Date of Admission", "Hospital", "Score", "Comments", "URL"])
    writer.writerows(posts)

print(f"✅ Saved {len(posts)} healthcare-related posts to {csv_filename}")

# Provide a download link for the CSV file
from google.colab import files
files.download(csv_filename)
```

```python
# Load dataset using pandas
file_path = "/content/reddit_healthcare_data.csv"
df = pd.read_csv(file_path)

# Create Great Expectations context
context = gx.get_context()

# Define a Pandas data source
data_source = context.data_sources.add_pandas("pandas")

data_asset = data_source.add_dataframe_asset(name="reddit_healthcare_asset")

# Define batch for the dataframe
batch_definition = data_asset.add_batch_definition_whole_dataframe("batch definition")
batch = batch_definition.get_batch(batch_parameters={"dataframe": df})

# Define Expectations
expectations = [
    gx.expectations.ExpectTableRowCountToEqual(value=100),  # At least 50 rows
    gx.expectations.ExpectColumnValuesToNotBeNull(column="Medical Condition"),  # No null values in Medical Condition
    gx.expectations.ExpectColumnPairValuesAToBeGreaterThanB(column_A="Score", column_B="Comments"),  # Score >= Comments
    gx.expectations.ExpectTableColumnsToMatchOrderedList(
        column_list=["Medical Condition", "Doctor", "Date of Admission", "Hospital", "Score", "Comments", "URL"]
    ),  # Ensure correct column names & order
    gx.expectations.ExpectColumnValuesToBeBetween(column="Score", min_value=0, max_value=10000)  # Score within a reasonable range
]

# Validate the dataset
validation_results = [batch.validate(expectation) for expectation in expectations]


```

```python
# Load dataset using pandas
file_path = "/content/healthcare_dataset.csv"
df = pd.read_csv(file_path)

# Create Great Expectations context
context = gx.get_context()

# Define a Pandas data source
data_source = context.data_sources.add_pandas("pandas")

data_asset = data_source.add_dataframe_asset(name="reddit_healthcare_asset")

# Define batch for the dataframe
batch_definition = data_asset.add_batch_definition_whole_dataframe("batch definition")
batch = batch_definition.get_batch(batch_parameters={"dataframe": df})

# Define Expectations
expectations = [

    gx.expectations.ExpectColumnToExist(column="age"),  # Ensure "age" column exists
    gx.expectations.ExpectColumnValuesToBeBetween(column="age", min_value=0, max_value=90),  # Age should be between 0 and 90
    gx.expectations.ExpectColumnValuesToNotBeNull(column="bmi"),  # No null values in BMI
    gx.expectations.ExpectColumnPairValuesToBeInSet(
        column_A="age",
        column_B="bmi",
        value_pairs_set={(25, 18.5), (40, 25.0), (60, 30.0)}
    )
]

# Validate the dataset
validation_results = [batch.validate(expectation) for expectation in expectations]
```
